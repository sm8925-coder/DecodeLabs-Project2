Data Loading & Structure Audit

```python
import pandas as pd
import numpy as np

df = pd.read_excel("/content/Dataset for Data Analytics.xlsx")

print("--- PHASE 1: DATA LOADING & STRUCTURE AUDIT ---")
print("Dataset Shape (Rows, Columns):", df.shape)
print("\nVariables Data Types:")
print(df.dtypes)
print("\nFirst 3 Records Preview:")
print(df.head(3))
```

Descriptive Statistics Calculations

```python
print("--- PHASE 2: DESCRIPTIVE STATISTICS CALCULATIONS ---")

numeric_cols = ['Quantity', 'UnitPrice', 'ItemsInCart', 'TotalPrice']

for col in numeric_cols:
    print(f"\n[Statistical Metrics for: {col}]")
    print(f"  - Total Count: {df[col].count()}")
    print(f"  - Mean Average: {round(df[col].mean(), 2)}")
    print(f"  - Median Value: {round(df[col].median(), 2)}")
    print(f"  - Standard Deviation: {round(df[col].std(), 2)}")
    print(f"  - Minimum Value: {df[col].min()}")
    print(f"  - Maximum Value: {df[col].max()}")
```

Trends & Categorical Patterns Performance

```python
print("--- PHASE 3: TRENDS & CATEGORICAL PATTERNS PERFORMANCE ---")

product_summary = df.groupby('Product').agg(
    Total_Revenue=('TotalPrice', 'sum'),
    Total_Quantity=('Quantity', 'sum'),
    Average_Order_Value=('TotalPrice', 'mean')
).round(2).sort_values(by='Total_Revenue', ascending=False)

print("\n1. Top Performing Products by Revenue:")
print(product_summary)

payment_summary = df.groupby('PaymentMethod').size().sort_values(ascending=False)
print("\n2. Payment Method Preferences:")
print(payment_summary)

referral_summary = df.groupby('ReferralSource').agg(
    Total_Sales=('TotalPrice', 'sum'),
    Order_Count=('OrderID', 'count')
).round(2).sort_values(by='Total_Sales', ascending=False)

print("\n3. Marketing Channels Efficiency (Referral Sources):")
print(referral_summary)
```

Outlier Detection

```python
print("--- PHASE 4: OUTLIER DETECTION (IQR METHOD) ---")

Q1 = df['TotalPrice'].quantile(0.25)
Q3 = df['TotalPrice'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df['TotalPrice'] < lower_bound) | (df['TotalPrice'] > upper_bound)]

print(f"Statistical Bounds for TotalPrice:")
print(f"  - Lower Bound threshold: {round(lower_bound, 2)}")
print(f"  - Upper Bound threshold: {round(upper_bound, 2)}")
print(f"\nExecution Result:")
print(f"  - Total Outliers Detected: {len(outliers)}")

if len(outliers) > 0:
    print("\nSample of Detected Outliers (Top 5 highest anomalous entries):")
    print(outliers[['OrderID', 'CustomerID', 'Product', 'TotalPrice']].sort_values(by='TotalPrice', ascending=False).head(5))
    print("\nVerification Gate: PASS - Outliers successfully mapped and isolated.")
else:
    print("\nVerification Gate: PASS - No extreme outliers detected within the 1.5*IQR boundary.")
```

Business Intelligence Summary & Export

```python
print("--- PHASE 5: BUSINESS INTELLIGENCE SUMMARY & EXPORT ---")

with pd.ExcelWriter('Exploratory_Data_Analysis_Project2.xlsx') as writer:
    df.describe().round(2).to_excel(writer, sheet_name='Descriptive_Stats')
    product_summary.to_excel(writer, sheet_name='Product_Trends')
    referral_summary.to_excel(writer, sheet_name='Marketing_Channels')
    if len(outliers) > 0:
        outliers.to_excel(writer, sheet_name='Detected_Outliers', index=False)

print("File generated successfully: Exploratory_Data_Analysis_Project2.xlsx")
print("\n--- ALL VERIFICATION GATES: PASS ---")
```
